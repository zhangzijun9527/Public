```python
a2=np.arange(24).reshape(4,6)
a2
```

```python
print('>',a2>12)
print('<',a2<12)
print('>=',a2>=10)
print('<=',a2<=10)
print('!=',a2!=10)
print('&',(a2>5) & (a2<10))
print('|',(a2>5) | (a2<10))
```

```python
print('>',a2[a2>12])
print('<',a2[a2<12])
print('>=',a2[a2>=10])
print('<=',a2[a2<=10])
print('!=',a2[a2!=10])
print('&',a2[(a2>5) & (a2<10)])
print('|',a2[(a2>20) | (a2<10)])
```
