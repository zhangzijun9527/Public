```python
import re
string = "hellow 123 world 456"
#  创建正则表达式
pattern = r"hellow"   #     \d+   匹配一个或多个数字
# 1.match 匹配开始位置 没有匹配成功返回一个空对象  匹配成功返回一个 re.match
match = re.match(pattern, string)
print(match)
# 2. search 匹配整个字符串 但是返回第一个匹配结果
search = re.search(pattern, string)
print(search.group())      #  .group 返回匹配对象
# 3.findall 匹配整个字符串 返回匹配到的结果
pattern = r'\d+'
findall = re.findall(pattern, string)
print(findall)
# 4.split       分割
split = re.split(pattern, string)
print(split)
# 5.sub 替换
sub = re.sub(pattern, "520", string)
print(sub)
string = '2026-10-20'
pattern = r'(\d+)-(\d+)-(\d+)'
match = re.match(pattern, string)
print(match.group())
print(match.group(2))
```
