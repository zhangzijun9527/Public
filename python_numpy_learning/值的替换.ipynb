```python
import numpy as np
```

```python
# 1.将np,arange(10)数组中的奇数全部都替换成-1.
a1 = np.arange(10)
a1[a1%2==1]=-1
a1
```

```python
# 2.有⼀个4⾏4列的数组（⽐如：np.random.randint(0,10,size=(4,4)),请将其中对⻆线的数取出来形成⼀
# 个⼀维数组，提示(使⽤np.eye)·
a2 = np.random.randint(0,10,size=(4,4))
print(a2)
arr = np.eye(4,dtype=bool)
a2[arr]
```

```python
# 3.有⼀个4⾏4列的数组，请取出其中(0,0),(1,2),(3,2)的点。
a3 = np.random.randint(1,10,size=(4,4))
print(a3)
a3[[0,1,3],[0,2,2]]
```

```python
# 4.有⼀个4⾏4列的数组，请取出其中2:3⾏(包括第3⾏)的所有数据。
a4=np.random.randint(0,10,size=(4,4))
print(a4)
a4[1:3,]
```

```python
# 5.有⼀个8⾏9列的数组，请将其中1-6⾏(包含第5⾏)的第8列⼤于3的数全部都取出来。
a5=np.random.randint(0,10,size=(8,9))
print(a5)
a5[1:6,8][a5[1:6,8]>3]
```

```python
# 6.替换数组中所有⼩于平均值的元素
a6=np.random.randint(0,10,size=(4,4))
print(a6)
arr = a6.mean()
print(arr)
a6[a6<arr] = arr
print(a6)
```

```python
# 7.将⼀个数组的边框元素设置为0
a7=np.random.randint(0,10,size=(4,4))
print(a7)
a7[:,(0,3)]=0
a7[(0,3),:]=0
a7
```

```python
# 8.反转⼆维数组的⾏
a8=np.random.randint(0,10,size=(4,4))
print(a8)
print(a8[::-1])
print(a8[:,::-1])
```

```python
# 9. 将3D数组的最后⼀个维度进⾏求和\
a9 = np.random.randint(0,10,size=(3,4,5))
print(a9)
arr=a9.sum(axis=-1)[2]
print(arr)
```
