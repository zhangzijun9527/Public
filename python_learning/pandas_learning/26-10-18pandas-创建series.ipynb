```
a = np.array([[1,2],[3,4]])
b = np.array([[5,6],[7,8]])
print(a)
print(b)
a1 = np.vstack((a,b))
a2 = np.hstack((a,b))
print(a1)
print(a2)
a3 = a.T
print(a3)
a4 = a.flatten()
print(a4)
a5= np.split(a,2)
print(a5)
```

```
data = np.array([1,2,3,4,5])
print(data)
a1 = pd.Series(data,index=['a','b','c','d','e'])
print(a1)
# 列表和数组索引和值必须一一对应
```

```
data = {
    'a':[1,2,3],
    'b':[4,5,6],
    'c':[7,8,9]
}
print(data)
a2 = pd.Series(data,index=['a','b','c','d'])
print(a2)
print(a2.values)
print(a2.index)
# 当字典索引超出时，返回nan空
```
