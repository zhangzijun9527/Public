```
import random
students = pd.read_excel(r'student.xlsx')
print(students)
print(students.columns)
```

```
# 1. 观察结构，调整列索引顺序为(id,学⽣id,学⽣姓名,年龄,性别,语⽂成绩,数学成绩,英语成绩,学⽣是否
# 毕业,学⽣是否休学)
df = students.reindex(columns=['id','学生id','学生姓名','年龄','性别','语文成绩','数学成绩','英语成绩','学生是否毕业','学生是否休学'])
print(df)
```

```
# 2. 增加⼀列关于学⽣状态(可使⽤1代码 在 0代表不在)
df['学生状态'] = df['学生是否毕业'].apply(lambda x: 1 if x == '是' else 0)
print(df)
```

```
# 3. 删除id这⼀列
del df['id']
print(df)
```

```
# 4. 查找出学⽣语⽂成绩⼤于90分的并输出的时候只有学⽣id-语句成绩
df = df[df['语文成绩']>90]
print(df.loc[:,'学生id':'语文成绩'])
```

```
data = {
    '班级': ['一班', '一班', '二班', '二班', '一班'],
    '姓名': ['张三', '李四', '王五', '赵六', '钱七'],
    '分数': [85, 92, 78, 88, 95]
}
df = pd.DataFrame(data)
print(df)
```

```
print("各班平均分：\n",df.groupby('班级')['分数'].mean())
print('\n各班统计：\n',df.groupby('班级')['分数'].agg(['sum','count']))
```
